* Fill in the blank in the HTML code string `html` to assign a `class` attribute to the second `<div>` element which has the value `"What-a-nasty-course"`.

```html
<html>
  <body>
    <div class="class1" id="div1">
```

````python
# Here this functions is written to display the exercise

from scrapy.selector.unified import Selector

def whats_my_class(html):
    sel = Selector(text=html)
    try:
        print( "The class you assigned to the first div element is:", sel.xpath('//div')[0].xpath('./@class' )[0].extract())
        print( "The class you assigned to the first p element is:", sel.xpath('//p')[0].xpath('./@class' )[0].extract())
        print( "The class you assigned to the second div element is:", sel.xpath('//div')[1].xpath('./@class' )[0].extract())
    except:
        print("No second div element class found!")

# HTML code string
# HTML code string
html = '''
<html>
  <body>
    <div class="class1" id="div1">
      <p class="class2">Visit DataCamp!</p>
    </div>
    <div class="What-a-nasty-course">
      <p class="class2">Keep up the good work!</p>
    </div>
  </body>
</html>
'''
# Print out the class of the second div element
whats_my_class(html)
````

Considering the following HTML:

```html
<html>
  <head>

    <link rel="stylesheet" type="text/css" href="style.css">
  </head>
  <body>
    <div class="class1" id="div1">
      <p class="class2">
        Visit <a href="http://datacamp.com/">DataCamp</a>!
      </p>
    </div>
    <div class="class1 class3" id="div2">

        Or search for it on <a href="http://www.google.com">Google</a>!
      </p>
    </div>
  </body>
</html>
```
- Moving from top to bottom, we see that the first `<a>` has an `href` attribute with the URL "http://datacamp.com/".
- This above `<a>` is the first child of the first `<p>` whose href attribute is the URL "http://datacamp.com/".
- A grandchild of a `<div>` element of `id="div1"` is the `<a>` with `href="http://datacamp.com/"`

Both `div` elements belong to the class `class1` (even though the second div element also belongs to `class3`), and both have an `<a>` element as a descendant (in this case, a grandchild). So, this direction points to two href attributes, the first is the `"http://datacamp.com/"`, the second is `"http://www.google.com"`.

>> It seems you need to be familiar not only with attributes `id=`, `class=`, `href=`, but also the notion of child, first child, second child, grandchild, parent, grandparent.



# **Crash Course in XPath**

## **Examples**

* `xpath = '/html/body/div[2]'`
  * `/` moves forward ONE generation: think of the tag-names as the "directory" names of folders on your computer.
  * `//` looks through all future generation
  * `[]` after a tag selects which siblings to choose
    * E.g. From `'/html/body/div[2]'`, there could be serveral `<div>` elements which are children of `<body>`.
    - `div[2]` means the second child (from above) AMONG THE `<div>` of the `<body>` as highlighted below.
    - The first child of the `<body>` is `<span>`.

image.png

* Direct to all `<table>` elements within the HTML code:
  * `xpath = '//table'`, or we could be more specific as
* Direct to all `<table>` elements which are descendants of the 2nd `<div>` child of `<body>`
  * `xpath = '/html/body/div[2]//table`

#### ***Exercise: A classy span***

Although we haven't yet gone deep into XPath, one thing we can do is select elements by their attributes using an XPath. For example, if we want to direct to the `<div>` element within the HTML document whose `id` attribute is `"uid"`, then we could write the XPath string `'//div[@id="uid"]'`. The first part of this string, `//div`, first looks at all `<div>` elements in the HTML document. Then, using the brackets, we specify that we want only the `<div>` element with a specific `id` attribute (in this case `uid`). To note, the phrase `@id="uid"` in the brackets would be read as "attribute `id` equals `uid`".

You will select all `span` elements whose class attribute equals `"span-class"`. (Note: span is just another possible tag-name). You do not need to see the actual HTML code to do this!

````python
# Be mindful of the use of two types of quotation marks here

xpath = "//span[@class='span-class']"
````

#### ***Exercise: Counting Elements in the Wild***

* `"/html/body/*"` selects all elements one generation below the body element without concern of the tag type
* `"/html/body//*"` selects all elements from all future generations of the body element (that is, all descendants of the body) regardless of tag type.
- The number of elements selected by the XPath string xpath = `"/*"` is equal to the number of root elements within the HTML document, which is typically the 1 html root element.
- The number of elements selected by the Xpath string `xpath = "//*"` is equal to the total number of elements in the entire HTML document.

# **(At)tribute**

* `@` in XPath represents "attribute"
  * `@class`
  * `@id`
  * `@href`
- xpath = `'//p'` directo to all `<p>`
- xpath = `'//p[@class="class-1"]'` directo to all `<p>` whose `class` is `class-1` as shown in the figure below.


image.png
- xpath = `'//*'` direct to all elements
- xpath = `'//*[@id="uid"]'` direct to all elements whose id equals "uid".
image.png
- xpath = `'//div[@uid="uid"]/p[2]'` navigate to any `<div>` whose id is "uid" and direct to the second `<p>` of that `<div>` element.

## **Selecting attribute**

In XPath  `.../@attribute_name` means choose the (value) of that attribute of the elements that come prior to  `/@attribute_name` (the element in `...`)

For more examples:




````python
from scrapy import Selector

def print_attribute(xpath):
    for i, el in enumerate(sel.xpath(xpath).extract()):
        print("%s" % (el))

html = '<a id="link1" class="menu" href="https://example.com"> Example </a>'
xpaths = ['//a/@class', '//a/@id', '//a/@href']
sel = Selector(text=html)
for xpath in xpaths:
    print_attribute(xpath)
````

#### ***Exercise: XPath Chaining***

`Selector` and `SelectorList` objects allow for chaining when using the `xpath` method. What this means is that you can apply the `xpath` method over once you've already applied it. For example, if `sel` is the name of our `Selector`, then

```
sel.xpath('/html/body/div[2]')
```
is the same as

* `sel.xpath('/html').xpath('./body/div[2]')`
* `sel.xpath('/html').xpath('./body').xpath('./div[2]')`

The only catch is that you need to "glue together" the XPath pieces by using a `.` (period) at the start of each subsequent XPath string.

#### ***Instructions***

* Chain together two xpath calls which result in the same selection as `sel.xpath('//div/span/p[3]')`



````python
from scrapy import Selector
html = '\n<html>\n<body>\n<div>HELLO</div>\n<div><p>GOODBYE</p></div>\n<div><span><p>NOPE</p><p>ALMOST</p><p>YOU GOT IT!</p></span></div>\n</body>\n</html>\n'
sel = Selector(text=html)

# Chain together xpath methods to select desired p element
print(sel.xpath('//div/span/p[3]').extract())
print(sel.xpath( '//div' ).xpath('./span/p[3]').extract())
````

# **From XPath  to  CSS**

* `/` is replaced by `>` (except first character)
  * **XPath**: `/html/body/div`
  * **CSS**: `html > body > div`
* `//` is replaced by a blank space (except first character)
  * **XPath**: `//div/span//p`
  * **CSS**: `div > span p`
* `[N]` is replaced by `:nth-of-type(N)`
  * **XPath**: `//div/p[2]`
  * **CSS**: `div > p:nth-of-type(2)`

## **Examples**

**XPath**
```text
xpath = '/html/body//div/p[2]'`
```

**CSS**

```
css = html > body div > p:nth-of-type(2)`
```

## **Attribites in CSS**

By <ins>**class**</ins>: Follow a tag name by `.` and the class name, e.g.

* `p.class-1` selects all `<p>` with `class="class-1"`

By <ins>**id**</ins>:  use a pound sign `#`, e.g.,

* `div#uid` selects all `<div>` with `id="uid"`

## **Examples**

* css = `'div#uid > p.class1'`

First navigates to the `<div>` element whose `id` is `"uid"`, then down one generation, and finally to whichever `<p>` element within that generation has a class attribute which belongs to `class1`.

* `css = '.class1'`

Select all html tag with `class="class1"`

## **Selector with CSS**

Use `.css()` method instead of `.xpath()`

````python
from scrapy import Selector

html = '''
<html>
  <body>
    <div class="hello datacamp">
      <p>Hello World!</p>
    </div>
    <p>Enjoy DataCamp!</p>
  </body>
</html>
'''
sel = Selector(text=html)
print(' '.join(sel.css("html > body > div > p").xpath('./text()').extract()))
print(' '.join(sel.xpath('/html/body/div/p').xpath('./text()').extract()))
print(' '.join(sel.css("div > p").xpath('./text()').extract()))
print(' '.join(sel.xpath('//div/p').xpath('./text()').extract()))
````

#### ***Exercise: The CSS Wildcard***

You can use the wildcard `*` in CSS Locators too! In fact, we can use it in a similar way, when we want to ignore the tag type. For example:

* The CSS Locator string `'*'` selects all elements in the HTML document.
* The CSS Locator string `'*.class-1'` selects all elements which belong to `class-1`, but this is unnecessary since the string `'.class-1'` will also do the same job.
* The CSS Locator string `'*#uid'` selects the element with `id` attribute equal to `uid`, but this is unnecessary since the string `'#uid'` will also do the same job.

In this exercise, we want you to work by analogy with the wildcard character you know from `XPath` notation to discover how to select all the children of a certain element in `CSS` Locator notation.

* Assign to the variable `css_locator` a CSS Locator string which will select all children (regardless of tag-type) of the unique element in the HTML document that has its `id` attribute equal to `uid`.

````python
# Create the CSS Locator to all children of the element whose id is uid
css_locator = '#uid > *'
xpath = '//*[@id="uid"]/*'
````

# **Attribute and Text Selection**
(For reference scroll up to see "**Selecting attribute**" section)

* Using <ins>**XPath**</ins>: `<xpath-to-element>/@attr-name` to select the attribute in XPath

    * `xpath = '//div[@id="uid"]/a/@href'`

* Using <ins>**CSS**</ins> Locator: `<css-to-element>::attr(attr-name)`

    * `css = 'div#uid > a::attr(href)'`



## **Text Extraction in Xpath**


```html
<p id="p-example">
    Hello world!
    Try <a href="http://www.datacamp.com">DataCamp</a> today!
</p>
```

* In <ins>**XPath**</ins> use `/text()` for only direct children and `//text()` for all descendants
  * Only children:  `/text()`
    * `sel.xpath('//p[@id="p-example"]/text()').extract()`
  * All descendants: `//text()`
    * `sel.xpath('//p[@id="p-example"]//text()').extract()`




````python
from scrapy import Selector

html = '''
<p id="p-example">
 Hello world!
 Try <a href="http://www.datacamp.com">DataCamp</a> today!
</p>'''
sel = Selector(text=html)

# In Xpath, use /text() for only texts in tag's children
print(sel.xpath('//p[@id="p-example"]/text()').extract())

# Use //text() for all texts in tag's all descendants
print(sel.xpath('//p[@id="p-example"]//text()').extract())

# Try BY MYSELF getting hyperlink
print(sel.xpath('//a/@href').extract())
````

## **Text Extraction in CSS**
(Be careful of a pesky ONE BLANK SPACE)

* For <ins>**CSS Locator**</ins>: Use `::text` for only children and
`   ::text` (have one blank space) for all descendants.
  * Only children
    * `sel.css('p#p-example::text').extract()`
  * All descendants: Has one blank space before `::`
    * `sel.css('p#p-example ::text').extract()`

````python
from scrapy import Selector

html= '''
<p id="p-example">
 Hello world!
 Try <a href="http://www.datacamp.com">DataCamp</a> today!
</p>'''

sel = Selector(text=html)

# Only children
print(sel.css('p#p-example::text').extract())

# All descendants
print(sel.css('p#p-example ::text').extract())

# Try BY MESELF getting hyperlink
print(sel.css('a::attr(href)').extract())
````

#### ***Exercise: Top Level Text***

This exercise will have you write an `XPath` and `CSS` Locator string to direct to the text of a specific paragraph `<p>` element. The `<p>` element in the HTML is uniquely defined by its `id` attribute, which is `"p3"`. With this small piece of information, you should be able to create the desired strings; however, we have preloaded the variable html with a string containing the HTML in which this link belongs, if you want to peruse it.

In this exercise, you will only be selecting the text within the element, which does not include the text in future generations of the element. We have created a function `print_results` for you to compare which elements your strings direct to.

* Assign to the variable `xpath` an XPath string directing to the text within the paragraph `<p>` element with `id` equal to `p3`, which does not include the text of future generations of this `<p>` element.
* Assign to the variable `css_locator` a CSS Locator string directing to this same text.

````python
from scrapy import Selector

def print_results( xpath, css_locator ):
    print( "Your XPath extracts to following:")
    print(sel.xpath(xpath).extract())
    print("_________________\n")
    print( "Your CSS Locator extracts the following:")
    print(sel.css(css_locator).extract())
    return None

html = '\n<html>\n<body>\n<div id="this-div">\n<p id="p1" class="class-1">This is not the element you are looking for</p>\n<p id="p2" class="class-12">\n<a href="https://www.google.com">Google</a> is linked to here, but this isn\'t the link you are looking for. \n</p>\n<p id="p3" class="class-1 class-12">\nHere is the <a href="https://www.datacamp.com" id="a-exercise">DataCamp</a> link you want!\n</p>\n</div>\n</body>\n</html>\n'
sel = Selector(text=html)
# Create an XPath string to the desired text.
xpath = '//p[@id="p3"]/text()'

# Create a CSS Locator string to the desired text.
css_locator = 'p#p3::text'

# Print the text from our selections
print_results( xpath, css_locator )
````

* Assign to the variable `xpath` an XPath string directing to the text within the paragraph `<p>` element with `id` equal to `p3`, which includes the text of future generations of this `<p>` element.

````python
from scrapy import Selector

def print_results( xpath, css_locator ):
    print( "Your XPath extracts to following:")
    print(sel.xpath(xpath).extract())
    print("_________________\n")
    print( "Your CSS Locator extracts the following:")
    print(sel.css(css_locator).extract())
    return None

html = '\n<html>\n<body>\n<div id="this-div">\n<p id="p1" class="class-1">This is not the element you are looking for</p>\n<p id="p2" class="class-12">\n<a href="https://www.google.com">Google</a> is linked to here, but this isn\'t the link you are looking for. \n</p>\n<p id="p3" class="class-1 class-12">\nHere is the <a href="https://www.datacamp.com" id="a-exercise">DataCamp</a> link you want!\n</p>\n</div>\n</body>\n</html>\n'
sel = Selector(text=html)
# Create an XPath string to the desired text.
xpath = '//p[@id="p3"]//text()'

# Create a CSS Locator string to the desired text.
css_locator = 'p#p3 ::text'

# Print the text from our selections
print_results(xpath, css_locator)
````

# **Response Object**

* has similar methods as those from **`Selector`**, but also collect  URL of the web that downloads html in the attribute
  * `response.url`
* Can move from one webstie to another by **`.follow()`** method

Chaining works like a Selector
- `response.xpath('//div').css('span.bio')`

Data extraction similar to Selector
- `response.xpath('//div/span[@class="bio"]').extract()`
`response.xpath('//div/span[@class="bio"]').extract_first()`

# **Spider start_requests method**

* Yield `scrapy.Request` object
* `scrapy.Request` send HTML from the `url` in `urls` to the `scrapy.response` object to the `parse` method in the `callback` argument.
* The 2nd variable `response` ใน `parse` method comes from `start_request` method (i.e., from `scrapy.Request`)

````python
def start_requests(self):
    urls = ["https://www.datacamp.com", "https://scrapy.org"]
    for url in urls:
        yield scrapy.Request(url=url, callback=self.parse)

def parse(self, response):
    html_file = "DC_Course.html"
    with open(html_file, "wb") as fout:
        fout.write(response.body)
````

#### ***Exercise: Self Referencing is Classy***

You probably have noticed that within the spider class, we always input the argument `self` in the `start_requests` and `parse` methods (just look in the sample code in this exercise!). This allows us to reference between methods within the class. That is, if we want to refer to the method `parse` within the `start_requests` method, we would need to write `self.parse` rather than just `parse`; what writing `self` does is tell the code: "Look in the same class as `start_requests` for a method called `parse` to use."

#### ***Exercise Crawler Time***

This will be your first chance to play with a spider which will crawl between sites (by first collecting links from one site, and following those links to parse new sites). This spider starts at the shortened DataCamp course directory, then extracts the links of the courses in the `parse` method; from there, it will follow those links to extract the course descriptions from each course page in the `parse_descr` method, and put these descriptions into the list `course_descrs`. Your job is to complete the code so that the spider runs as desired!

We have created a function `inspect_spider` which will print out one of the course descriptions you scrape (if done correctly)!

* Fill in the two blanks (one in each of the parsing methods) with the appropriate entries so that the spider can move from the first parsing method to the second correctly.

````python
import scrapy
import requests
from scrapy.http import TextResponse

'https://assets.datacamp.com/production/repositories/2560/datasets/19a0a26daa8d9db1d920b5d5607c19d6d8094b3b/all_short'

def print_first_descr(course_descrs):
    print("The first course description is: \n", course_descrs[0])

def inspect_spider(s):
    news = s()
    try:
        req1 = list(news.start_requests())[0]
        html1 = requests.get(req1.url).content
        response1 = TextResponse(url=req1.url, body=html1, encoding='utf-8' )
        req2 = list(news.parse(response1 ))[0]
        html2 = requests.get(req2.url).content
        response2 = TextResponse(url=req2.url, body=html2, encoding='utf-8' )
        for d in news.parse_descr(response2):
            print("One course description you found is:", d )
            break
    except:
        print("Oh no! Something is wrong with the code. Keep trying!")

# Create the Spider class
class DCdescr(scrapy.Spider):
  name = 'dcdescr'
  # start_requests method
  def start_requests(self):
    yield scrapy.Request(url=url_short, callback=self.parse)

  # First parse method
  def parse(self, response):
    links = response.css('div.course-block > a::attr(href)').extract()
    # Follow each of the extracted links
    for link in links:
      yield response.follow(url=link, callback=self.parse_descr)

  # Second parsing method
  def parse_descr(self, response):
    # Extract course description
    course_descr = response.css('p.course__description::text').extract_first()
    # For now, just yield the course description
    yield course_descr


# Inspect the spider
inspect_spider(DCdescr)
````

# **Scraping for real**

*  ต้องไปส่องไฟล์ html ต้นฉบับแล้วดูว่าสิ่งที่เราจะเอา มันอยู่ภายใต้ tags ที่มีข้อมูลแบบไหน
* ในตัวอย่างข้างล่าง ต้องการเก็บชื่อคอร์ส (course title) จากการส่องไฟล์ดูพบว่า title และ chapter title อยู่ภายใต้ tag
  * `<h1 class="title ..."> `(ใช้ `contains` หรือ `css` จะหาง่ายสุด)
  * `<h4 class="chapter__title"> `

````python
import scrapy, requests
from scrapy.crawler import CrawlerProcess



class DC_Chapter_Spider(scrapy.Spider):

    name = "dc_chapter_spider"

    def start_request(self):
        url = "https://www.datacamp.com/courses/all"
        yield scrapy.Request(url=url, callback=self.parse_front)

    def parse_front(self, response):
        # Narrow in on the course blocks
        course_blocks = response.css('div.course-block')
        # Direct to the course links
        course_links = course_blocks.xpath('./a/@href')
        # Extract the links as a list of strings
        links_to_follow = course_links.extract()
        for link in links_to_follow:
            yield response.follow(url=url, callback=self.parse_pages)

    def parse_pages(self, response):
        # Direct to course title's text
        crs_title = response.xpath('//h1[contains(@class, "title")]/text()')
        # Extract and clean the course title's text
        crs_title_ext = crs_title.extract_first().strip()
        # Direct to the chapter text
        ch_titles = response.css('h4.chapter__title::text')
        # Extract and clean up
        ch_titles_ext = [t.strip() for t in ch_titles.extract()]
        # Store in dict
        dc_dict[crs_title_ext] = ch_titles_ext


dc_dict = dict()

process = CrawlerProcess()
process.crawl(DC_Chapter_Spider)
process.start()
````

#### ***Exercise: Time to Run***

* Fill in the one blank at the end of the `parse_pages` methods to assign the chapter titles to the dictionary whose key is the corresponding course title.

>> #### *Running this on colab is complicated and the codes on DataCamp is outdated that it cannot even show the output.*

>> #### *On top of this you need to restart runtime everytime you run the code below.*

````python
import scrapy, requests, logging, nest_asyncio
from scrapy.http import TextResponse
from scrapy.crawler import CrawlerProcess
from scrapy.crawler import AsyncCrawlerRunner

nest_asyncio.apply()



url_short = 'https://assets.datacamp.com/production/repositories/2560/datasets/19a0a26daa8d9db1d920b5d5607c19d6d8094b3b/all_short'

# Create the Spider class
class DC_Chapter_Spider(scrapy.Spider):
    name = "dc_chapter_spider"

    async def start(self):
        yield scrapy.Request(
            url=url_short,
            callback=self.parse_front
        )

    def parse_front(self, response):
        course_blocks = response.css('div.course-block')
        course_links = course_blocks.xpath('./a/@href')
        links_to_follow = course_links.extract()

        for url in links_to_follow:
            yield response.follow(
                url=url,
                callback=self.parse_pages
            )

    def parse_pages(self, response):
        crs_title = response.xpath(
            '//h1[contains(@class,"title")]/text()'
        )
        crs_title_ext = crs_title.extract_first().strip()

        ch_titles = response.css('h4.chapter__title::text')
        ch_titles_ext = [t.strip() for t in ch_titles.extract()]

        dc_dict[crs_title_ext] = ch_titles_ext

def previewCourses( dc_dict, n = 3 ):
  crs_titles = list( dc_dict.keys() )
  print( "A preview of DataCamp Courses:")
  print("---------------------------------------\n")
  for t in crs_titles[:n]:
    print( "TITLE: %s" % t)
    for i,ct in enumerate(dc_dict[t]):
      print("\tChapter %d: %s" % (i+1,ct) )
    print("")

# Initialize the dictionary **outside** of the Spider class
dc_dict = dict()

runner = AsyncCrawlerRunner({
    "TWISTED_REACTOR_ENABLED": False
})

await runner.crawl(DC_Chapter_Spider)

# Run the Spider
logging.disable(logging.CRITICAL)
process = CrawlerProcess()
process.crawl(DC_Chapter_Spider)
process.start()

# Print a preview of courses
previewCourses(dc_dict)
````

Create a CSS Locator string direct to the text of the course description. All you need to know is that from the course page, the course description text is within a paragraph `<p>` element which belongs to the class `course__description`

* Fill in the one blank below in the `parse_pages` method with a CSS Locator string which directs to the text within the paragraph `<p>` element which belongs to the class `course__description`.

````python
import scrapy, requests, logging, nest_asyncio
from scrapy.http import TextResponse
from scrapy.crawler import CrawlerProcess
from scrapy.crawler import AsyncCrawlerRunner

nest_asyncio.apply()

url_short = 'https://assets.datacamp.com/production/repositories/2560/datasets/19a0a26daa8d9db1d920b5d5607c19d6d8094b3b/all_short'

def previewCourses(dc_dict, n = 1):
    crs_titles = list(dc_dict.keys())
    print( "A preview of DataCamp Courses:")
    print("---------------------------------------\n")
    for t in crs_titles[:n]:
        print("TITLE: %s" % t)
        print("\tDescription: %s" % dc_dict[t])
        print("")

# Create the Spider class
class DC_Description_Spider(scrapy.Spider):
    name = "dc_chapter_spider"
    # start_requests method
    async def start(self):
        yield scrapy.Request(
            url=url_short,
            callback=self.parse_front
        )
    # First parsing method
    def parse_front(self, response):
        course_blocks = response.css('div.course-block')
        course_links = course_blocks.xpath('./a/@href')
        links_to_follow = course_links.extract()
        for url in links_to_follow:
            yield response.follow(url=url, callback=self.parse_pages)
    # Second parsing method
    def parse_pages(self, response):
        # Create a SelectorList of the course titles text
        crs_title = response.xpath('//h1[contains(@class,"title")]/text()')
        # Extract the text and strip it clean
        crs_title_ext = crs_title.extract_first().strip()
        # Create a SelectorList of course descriptions text
        crs_descr = response.css('p.course__description::text')
        # Extract the text and strip it clean
        crs_descr_ext = crs_descr.extract_first().strip()
        # Fill in the dictionary
        dc_dict[crs_title_ext] = crs_descr_ext

# Initialize the dictionary **outside** of the Spider class
dc_dict = dict()

runner = AsyncCrawlerRunner({
    "TWISTED_REACTOR_ENABLED": False
})

await runner.crawl(DC_Description_Spider)

# Run the Spider
logging.disable(logging.CRITICAL)
process = CrawlerProcess()
process.crawl(DC_Description_Spider)
process.start()

# Print a preview of courses
previewCourses(dc_dict)
````

On the course directory page of DataCamp, each listed course has a title and a short course description. This spider will be used to scrape the course directory to extract the course titles and short course descriptions. You will not need to follow any links this time. Everything you need to know is:

* The course titles are defined by the text within an `<h4>` element whose class contains the string `block__title`.
* The short course descriptions are defined by the text within a paragraph `<p>` element whose class contains the string `block__description`

* Assign to the variable `crs_titles` the extracted list of course titles on the DataCamp course directory page. You should use the `contains` call within your `XPath`, and your `XPath` string should point to the `text` of the selected objects.

* Assign to the variable `crs_descrs` the extracted list of short course descriptions. You should use the `contains` call within your `XPath`. You should use the `contains` call within your `XPath`, and your `XPath` string should point to the `text` of the selected objects.

(Since we want a list of extracted data, we will use the `extract()` call (rather than `extract_first()`). )

````python
# parse method
def parse(self, response):
    # Extracted course titles
    crs_titles = response.xpath('//h4[contains(@class, "block__title")]/text()').extract()
    # Extracted course descriptions
    crs_descrs = response.xpath('//p[contains(@class, "block__description")]/text()').extract()
    # Fill in the dictionary: it is the spider output
    for crs_title, crs_descr in zip(crs_titles, crs_descrs):
        dc_dict[crs_title] = crs_descr
````
